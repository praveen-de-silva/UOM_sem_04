## Trip Duration (Metric / Continuous Data)

```python
# Feature: Trip Duration (Continuous Metric Data)
# UNSUITABLE: A Pie Chart is for a small number of categories. 
# There are thousands of different trip durations, so a pie chart would have too many invisible slices!
```

```python
# Feature: Trip Duration (Continuous Metric Data)
# UNSUITABLE: Tag clouds are used for text/words (Nominal Data). Trip duration is made of numbers.
```

```python
# Feature: Trip Duration (Continuous Metric Data)
# UNSUITABLE: Bar charts are for categorical data (like counting how many males vs females). 
# For continuous numbers, we use a Histogram instead.
```

```python
# Feature: Trip Duration (Continuous Metric Data)
# UNSUITABLE: A clustered or stacked chart requires a second category (like User Type) to stack on top of each other. 
# We are only looking at one feature right now.
```

```python
# Feature: Trip Duration (Continuous Metric Data)
# SUITABLE: Great for showing the Five-Number Summary and outliers!
plt.figure(figsize=(10, 2))
plt.boxplot(df['Trip_Duration_in_min'], vert=False)
plt.title('Box Plot of Trip Duration')
plt.xlabel('Minutes')
plt.show()
```

```python
# Feature: Trip Duration (Continuous Metric Data)
# SUITABLE: Perfect for showing the distribution of metric data. (We filter to < 60 mins to see the shape clearly).
plt.figure(figsize=(10, 5))
plt.hist(df[df['Trip_Duration_in_min'] < 60]['Trip_Duration_in_min'], bins=30, color='skyblue')
plt.title('Histogram of Trip Duration (Under 1 Hour)')
plt.xlabel('Minutes')
plt.ylabel('Frequency')
plt.show()
```

```python
# Feature: Trip Duration (Continuous Metric Data)
# SUITABLE: Shows the running total of rides as duration increases.
plt.figure(figsize=(10, 5))
plt.hist(df[df['Trip_Duration_in_min'] < 60]['Trip_Duration_in_min'], bins=30, cumulative=True, color='lightgreen')
plt.title('Cumulative Histogram of Trip Duration')
plt.xlabel('Minutes')
plt.ylabel('Cumulative Frequency')
plt.show()
```

## Start Time

```python
# Feature: Start Time (Ordered Discrete)
# UNSUITABLE: While technically possible, 24 slices in a Pie Chart is too messy to read. 
# Your lecture notes say "Pie Chart (small categories only)".
```

```python
# Feature: Start Time (Ordered Discrete)
# UNSUITABLE: Tag clouds are for text words, not numerical hours.
```

```python
# Feature: Start Time (Ordered Discrete)
# SUITABLE: Great for showing frequency per hour.
df['Start Hour'] = pd.to_datetime(df['Start Time']).dt.hour
start_hour_counts = df['Start Hour'].value_counts().sort_index()
plt.figure(figsize=(10, 5))
plt.bar(start_hour_counts.index, start_hour_counts.values, color='orange')
plt.title('Bar Chart: Rides per Start Hour')
plt.xlabel('Hour of the Day (0-23)')
plt.ylabel('Number of Rides')
plt.xticks(range(0, 24))
plt.show()
```

```python
# Feature: Start Time (Ordered Discrete)
# UNSUITABLE: We need a second feature (like 'Gender') to stack or cluster the lines. We only have one feature here.
```

```python
# Feature: Start Time (Ordered Discrete)
# SUITABLE: Perfect for ordered data over time!
# Make sure Start Hour exists (create it if needed)
if 'Start Hour' not in df.columns:
	df['Start Hour'] = pd.to_datetime(df['Start Time']).dt.hour

start_hour_counts = df['Start Hour'].value_counts().sort_index()
plt.figure(figsize=(10, 5))
plt.step(start_hour_counts.index, start_hour_counts.values, color='red', where='mid')
plt.title('Step Chart: Rides per Start Hour')
plt.xlabel('Hour of the Day (0-23)')
plt.ylabel('Number of Rides')
plt.xticks(range(0, 24))
plt.show()
```

```python
# Feature: Start Time (Ordered Discrete)
# UNSUITABLE: Hours cycle round from 23 back to 0. Finding the "median" hour using a box plot is mathematically confusing and not standard practice.
```

```python
# Feature: Start Time (Ordered Discrete)
# SUITABLE: Shows the distribution of rides across the 24-hour bins.
plt.figure(figsize=(10, 5))
plt.hist(df['Start Hour'], bins=24, color='gold', edgecolor='black')
plt.title('Histogram of Start Hour')
plt.xlabel('Hour of the Day (0-23)')
plt.ylabel('Frequency')
plt.xticks(range(0, 24))
plt.show()
```

```python
# Feature: Start Time (Ordered Discrete)
# SUITABLE: Shows how the total number of rides builds up from midnight to the end of the day.
plt.figure(figsize=(10, 5))
plt.hist(df['Start Hour'], bins=24, cumulative=True, color='brown', edgecolor='black')
plt.title('Cumulative Histogram of Start Hour')
plt.xlabel('Hour of the Day (0-23)')
plt.ylabel('Cumulative Frequency')
plt.xticks(range(0, 24))
plt.show()
```
